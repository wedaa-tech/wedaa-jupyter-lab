#### StandardScaler

Purpose:
Standardizes (normalizes) data so that each feature has:

#### mean=0,standard deviation=1

This helps models like SVMs, Neural Networks, or Gradient Boosting converge faster and perform better.

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

X = np.array([[10, 200],
              [20, 400],
              [30, 600]])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original X:\n", X)
print("\nScaled X:\n", X_scaled)
print("\nFeature Means:", scaler.mean_)
print("Feature Std Devs:", scaler.scale_)
```
