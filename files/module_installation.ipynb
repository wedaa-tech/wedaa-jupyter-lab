```python
import pyodide

# Get global JS context safely
window = pyodide.js.globalThis

async def copy_text(text: str):
    try:
        await window.navigator.clipboard.writeText(text)
        print("✅ Copied to clipboard successfully!")
    except Exception as e:
        print("⚠️ Clipboard access failed:", e)

async def paste_text():
    try:
        text = await window.navigator.clipboard.readText()
        print("📋 Clipboard content:")
        print(text)
        return text
    except Exception as e:
        print("⚠️ Clipboard read failed:", e)
        return None

```
