#### Sample Space and Events


Sample space (S): the set of all possible outcomes of an experiment.

Event (E): a subset of the sample space — one or more outcomes that share some property of interest.

Example

Suppose you’re analyzing whether a light bulb made in a factory is Defective (D) or Not Defective (N).

Sample space: S={D,N}

Event:
Let E={N} → “The bulb is not defective.”

If you randomly pick one bulb from production, the probability of event 

E is: 
𝑃(𝐸) = Number of Non-Defective Bulbs / Total Bulbs





#### 
Axioms of Probability (Kolmogorov’s Axioms)

These are the three rules that all probabilities must follow:

#### Non-Negativity
Probability of any event is ≥ 0  -> 𝑃(𝐴)≥0

#### Normalization
Probability of the sample space is 1 -> P(S)=1

#### Additivity
If A and B are mutually exclusive, 
Then,

P(A∪B)=P(A)+P(B)


#### From These Axioms, We Derive Important Propositions

**Proposition 1** — Probability of the Impossible Event
$$
P(\varnothing) = 0
$$
The probability of an impossible event (something that can’t happen) is zero.

**Proposition 2** - Range of Probability

$$
0 \leq P(A) \leq 1
$$
Probability always lies between 0 and 1 (inclusive).

**Proposition 3** - Complementary Events

$$
P(A') = 1 - P(A)
$$

If the probability of it raining is 0.3,
then the probability of not raining is  -> 1 − 0.3 = 0.7

**Proposition 4** - Addition Rule (General)

For any two events A and 𝐵:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

If events can happen together, you must subtract the overlap P(A∩B) to avoid double-counting.

**Proposition 5** - Conditional Probability

For two dependent events A and 𝐵:

The probability that event A happens given that B already happened.

$$
P(A|B) = \frac{P(A \cap B)}{P(B)}
$$


**Proposition 6** - Multiplication Rule
From the conditional probability definition:

$$
P(A \cap B) = P(A|B) \times P(B)
$$

The probability of both A and B happening = probability that B happens × probability that A happens after B.

**Proposition 7** - Total Probability Theorem
If 
𝐵1,𝐵2,𝐵3,…,𝐵𝑛  form a partition of the sample space:

$$
P(A) = \sum_{i=1}^{n} P(B_i) \, P(A|B_i)
$$
To find the total probability of an event A, sum over all possible ways A can happen.

**Proposition 8** - Bayes’ Theorem

For mutually exclusive and exhaustive events 𝐵1,𝐵2,…,𝐵𝑛 :

$$
P(B_j|A) = \frac{P(B_j) \, P(A|B_j)}{\sum_{i=1}^{n} P(B_i) \, P(A|B_i)}
$$

Helps reverse conditional probabilities — find P(cause∣effect) using P(effect∣cause).



#### Sample Spaces with Equally Likely Outcomes

Concept

If all outcomes in the sample space have the same chance of occurring, then:

P(E)= Number of outcomes in E / Total number of outcomes in S 


Example

You roll a fair die (6 sides): S={1,2,3,4,5,6}

Event 
E={2,4,6} → rolling an even number

P(E): 3/6 = 0.5



#### Commutative Law
Definition: Order of events doesn’t matter when taking their union or intersection:

A∪B = B∪A

A∩B = B∩A


#### Associative Law
Definition: When combining three or more events, the grouping (parentheses) doesn’t matter:

(A∪B)∪C=A∪(B∪C)

(A∩B)∩C=A∩(B∩C)



#### Distributive Law
Definition: Union and intersection distribute over each other — similar to how multiplication distributes over addition in arithmetic.

A∩(B∪C)=(A∩B)∪(A∩C)

A∪(B∩C)=(A∪B)∩(A∪C)

Meaning: When an event intersects or unions with a group of other events, you can “distribute” it across them.




### 🎲 Probability Propositions in Statistics

Probability theory is built upon a few basic rules called **axioms** (Kolmogorov’s Axioms), and many **derived propositions** that follow from them.

---

### 🧩 Axioms of Probability

### **Axiom 1 — Non-Negativity**

$$
P(A) \geq 0
$$

📘 The probability of any event is always **non-negative**.

---

### **Axiom 2 — Certainty**

$$
P(S) = 1
$$

📘 The probability of the **sample space** (the set of all possible outcomes) is **1**.

---

### **Axiom 3 — Additivity (Mutually Exclusive Events)**

If events \( A \) and \( B \) are **mutually exclusive** (cannot occur together):

$$
P(A \cup B) = P(A) + P(B)
$$

📘 The probability of either \( A \) or \( B \) occurring is the **sum of their probabilities**.

---

### 🧮 Derived Propositions

### **Proposition 1 — Probability of an Impossible Event**

$$
P(\varnothing) = 0
$$

📘 The probability of an **impossible event** is zero.

---

### **Proposition 2 — Range of Probability**

$$
0 \leq P(A) \leq 1
$$

📘 All probabilities must lie between **0 and 1**.

---

### **Proposition 3 — Complementary Events**

$$
P(A') = 1 - P(A)
$$

📘 The probability that event \( A \) does **not** occur is one minus the probability that it **does** occur.

---

### **Proposition 4 — Addition Rule (General)**

For any two events \( A \) and \( B \):

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

📘 Subtract the intersection \( P(A \cap B) \) to avoid double-counting when events can occur together.

---

### **Proposition 5 — Conditional Probability**

$$
P(A|B) = \frac{P(A \cap B)}{P(B)}
$$

📘 The probability that **A occurs given that B has occurred**.

---

### **Proposition 6 — Multiplication Rule**

$$
P(A \cap B) = P(A|B) \times P(B)
$$

📘 The probability of **both A and B** occurring equals the probability that **B happens** times the probability that **A happens given B**.

---

### **Proposition 7 — Total Probability Theorem**

If \( B_1, B_2, \ldots, B_n \) form a **partition** of the sample space:

$$
P(A) = \sum_{i=1}^{n} P(B_i) \, P(A|B_i)
$$

📘 The total probability of \( A \) can be found by summing across all possible disjoint events \( B_i \).

---

### **Proposition 8 — Bayes’ Theorem**

For mutually exclusive and exhaustive events \( B_1, B_2, \ldots, B_n \):

$$
P(B_j|A) = \frac{P(B_j) \, P(A|B_j)}{\sum_{i=1}^{n} P(B_i) \, P(A|B_i)}
$$

📘 Used to **reverse conditional probabilities**, e.g., finding \( P(\text{Cause}|\text{Effect}) \) given \( P(\text{Effect}|\text{Cause}) \).

---

### 🧾 Summary Table

| Proposition | Formula | Description |
|--------------|----------|-------------|
| Non-Negativity | \( P(A) \ge 0 \) | Probability can’t be negative |
| Certainty | \( P(S) = 1 \) | Whole sample space = 1 |
| Additivity | \( P(A ∪ B) = P(A) + P(B) \) | For mutually exclusive A, B |
| Complement | \( P(A') = 1 - P(A) \) | Probability of “not A” |
| Addition Rule | \( P(A ∪ B) = P(A) + P(B) - P(A ∩ B) \) | General case |
| Conditional | \( P(A|B) = \frac{P(A ∩ B)}{P(B)} \) | Probability of A given B |
| Multiplication | \( P(A ∩ B) = P(A|B)P(B) \) | Joint probability |
| Total Probability | \( P(A) = \sum P(B_i)P(A|B_i) \) | Combining all cases |
| Bayes’ Theorem | \( P(B_j|A) = \frac{P(B_j)P(A|B_j)}{\sum P(B_i)P(A|B_i)} \) | Reverse probability |

---

📘 **Tip:**  
In real-world problems, most probability questions rely on **these 8 propositions** — especially the **Addition Rule**, **Conditional Probability**, and **Bayes’ Theorem**.



#### Concept — Basic Principle of Counting

The Basic Principle of Counting (also called the Multiplication Rule) helps us find the total number of possible outcomes when a process is made up of two or more independent stages.

If an event (or experiment) can be performed in m different ways,
and after it, another event can be performed in n different ways,
then the total number of ways both events can occur is:

Total ways=m×n

Or:
If a process has k independent stages,
and each stage i can be done in 𝑛𝑖 ways, then:

$$
\text{Total number of possible outcomes} = n_1 \times n_2 \times n_3 \times \cdots \times n_k
$$


### 🧮 Generalised Basic Principle of Counting

### 📘 Statement
If a process consists of **k independent stages**,  
and each stage \( i \) can be performed in \( n_i \) different ways,  
then the **total number of ways** the entire process can occur is:

$$
\text{Total ways} = n_1 \times n_2 \times n_3 \times \cdots \times n_k
$$

Or, in compact form:

$$
N = \prod_{i=1}^{k} n_i
$$

---

### 🧠 Example 1 — Simple Case

- First digit can be chosen in **5 ways**  
- Second digit can be chosen in **3 ways**

Then:
$$
\text{Total outcomes} = 5 \times 3 = 15
$$

✅ There are 15 possible outcomes.

---

### 🧮 Example 2 — Book Selection

A student chooses:
- 1 Maths book out of 4  
- 1 Science book out of 3  
- 1 English book out of 5  

$$
\text{Total ways} = 4 \times 3 \times 5 = 60
$$

✅ 60 combinations possible.

---

### 🧮 Example 3 — Password Formation

A password has:
- 2 lowercase letters → \( 26^2 \)
- 3 digits → \( 10^3 \)

Then:

$$
\text{Total passwords} = 26^2 \times 10^3 = 6,760,000
$$

✅ There are **6,760,000** possible passwords.

---

### 💡 Key Idea
Each independent choice **multiplies** the total number of possibilities.
This principle is the **foundation of permutations and combinations**.




### 🎯 Basic Principles of Counting in Probability

Counting principles help us find **how many ways** an event or process can occur.  
There are two main rules:

---

### 🧩 1. Addition Principle of Counting

### 📘 Statement
If an event **A** can occur in **m** ways,  
and another event **B** can occur in **n** ways,  
and **A and B cannot occur together** (mutually exclusive),  
then the total number of ways **either A or B** can occur is:

$$
\text{Total ways} = m + n
$$

---

### 🧮 Example 1 — Choosing a Book

- You have **4 Maths books**  
- and **3 Science books**

If you want to select **one book** (either Maths or Science):

$$
\text{Total ways} = 4 + 3 = 7
$$

✅ There are **7 possible choices**.

---

### 🧮 Example 2 — Transportation

A student can:
- Go to school by **3 different bus routes**, or  
- By **2 different train routes**

Then:
$$
\text{Total ways} = 3 + 2 = 5
$$

✅ 5 total ways to reach school.

---

### 🧠 Key Idea
Use **Addition** when the choices are **either–or**.  
> One event happens **or** another happens (not both).

---

## 🧩 2. Multiplication (Generalised Basic) Principle of Counting

### 📘 Statement
If a process consists of **k independent stages**,  
and each stage \( i \) can be performed in \( n_i \) different ways,  
then the **total number of ways** the entire process can occur is:

$$
\text{Total ways} = n_1 \times n_2 \times n_3 \times \cdots \times n_k
$$

Or, in compact form:

$$
N = \prod_{i=1}^{k} n_i
$$

---

### 🧮 Example 1 — Simple Code Formation

- First digit can be chosen in **5 ways**  
- Second digit can be chosen in **3 ways**

Then:
$$
\text{Total outcomes} = 5 \times 3 = 15
$$

✅ There are **15 possible codes**.

---

### 🧮 Example 2 — Book Selection

A student chooses:
- 1 Maths book out of 4  
- 1 Science book out of 3  
- 1 English book out of 5  

Then:
$$
\text{Total ways} = 4 \times 3 \times 5 = 60
$$

✅ 60 combinations possible.

---

### 🧮 Example 3 — Password Formation

A password has:
- 2 lowercase letters → \( 26^2 \)
- 3 digits → \( 10^3 \)

Then:
$$
\text{Total passwords} = 26^2 \times 10^3 = 6,760,000
$$

✅ There are **6,760,000** possible passwords.

---

### 🧠 Key Idea
Use **Multiplication** when the choices are **and** conditions.  
> One event happens **and** another happens (in sequence).

---

## 🧾 Summary Table

| Principle | When Used | Formula | Keyword | Example |
|------------|------------|----------|----------|----------|
| **Addition** | Events are mutually exclusive | \( m + n \) | “**OR**” | Choosing a book (Maths or Science) |
| **Multiplication** | Events occur in sequence (independent) | \( n_1 \times n_2 \times \cdots \times n_k \) | “**AND**” | Selecting books from multiple subjects |

---

### 💡 Quick Tip
- **Addition → OR rule** → choose between options.  
- **Multiplication → AND rule** → perform sequential steps.  

These two are the **building blocks** for **permutations, combinations, and probability**.



### 🧮 Permutations and Combinations in Probability

Once we understand the **Addition** and **Multiplication Principles**,  
we can apply them to situations where we arrange or select objects.  
These are known as **Permutations** and **Combinations**.

---

### 🎯 1. Permutations (Arrangements)

### 📘 Definition
A **permutation** is an **arrangement** of objects in a specific **order**.  
Order **matters** in permutations.

> If you change the order, it becomes a different permutation.

---

### 🧩 Formula — Permutations of n distinct objects taken r at a time

$$
{}^nP_r = \frac{n!}{(n - r)!}
$$

Where:  
- \( n \) = total number of objects  
- \( r \) = number of objects to arrange  
- \( n! \) (n factorial) = \( n \times (n-1) \times (n-2) \times \cdots \times 1 \)

---

### 🧮 Example 1 — Simple Arrangement

> How many ways can 3 letters A, B, and C be arranged?

$$
{}^3P_3 = \frac{3!}{(3-3)!} = \frac{6}{1} = 6
$$

✅ Possible arrangements: **ABC, ACB, BAC, BCA, CAB, CBA**

---

### 🧮 Example 2 — Partial Arrangement

> How many ways can you arrange 3 letters out of 5 (A, B, C, D, E)?

$$
{}^5P_3 = \frac{5!}{(5-3)!} = \frac{120}{2} = 60
$$

✅ There are **60 possible arrangements**.

---

### 🧠 When to Use
Use **permutations** when **order matters**.  
> e.g. “Arrange”, “Seat”, “Rank”, “Position”, “Password (ordered)” etc.

---

## 🎯 2. Combinations (Selections)

### 📘 Definition
A **combination** is a **selection** of objects where **order doesn’t matter**.

> ABC, BAC, and CAB are all the same combination.

---

### 🧩 Formula — Combinations of n distinct objects taken r at a time

$$
{}^nC_r = \frac{n!}{r!(n - r)!}
$$

Where:  
- \( n \) = total number of objects  
- \( r \) = number of objects to choose  
- \( r! \) = factorial of r

---

### 🧮 Example 1 — Choosing Students

> How many ways can you choose 2 students from a group of 4?

$$
{}^4C_2 = \frac{4!}{2!(4-2)!} = \frac{24}{4} = 6
$$

✅ Possible pairs: **AB, AC, AD, BC, BD, CD**

---

### 🧮 Example 2 — Committee Formation

> A committee of 3 is to be formed from 5 members.

$$
{}^5C_3 = \frac{5!}{3!(5-3)!} = \frac{120}{12} = 10
$$

✅ There are **10 possible committees**.

---

### 🧠 When to Use
Use **combinations** when **order doesn’t matter**.  
> e.g. “Select”, “Choose”, “Form a team”, “Group”, “Committee” etc.

---

## 🔄 Relationship Between Permutations and Combinations

Permutations are just combinations **with arrangements**.

$$
{}^nP_r = {}^nC_r \times r!
$$

📘 Meaning:  
If you first **choose r items** (\( {}^nC_r \)) and then **arrange them** (\( r! \)),  
you get all possible **permutations**.

---

## 🧾 Summary Table

| Concept | Formula | Order Matters? | Example |
|----------|----------|----------------|----------|
| **Permutation** | \( {}^nP_r = \frac{n!}{(n-r)!} \) | ✅ Yes | Arranging letters, ranking players |
| **Combination** | \( {}^nC_r = \frac{n!}{r!(n-r)!} \) | ❌ No | Choosing committee members |

---

### 💡 Quick Tip

| Keyword | Use |
|----------|-----|
| **Arrange**, **Seat**, **Order**, **Password** | Permutation |
| **Select**, **Choose**, **Form**, **Pick**, **Team** | Combination |

---

### 🧠 Example Recap

| Scenario | Type | Formula | Result |
|-----------|------|----------|---------|
| Arrange 3 letters from 5 | Permutation | \( {}^5P_3 = 60 \) | 60 |
| Choose 2 players from 4 | Combination | \( {}^4C_2 = 6 \) | 6 |
| Form committee of 3 from 5 | Combination | \( {}^5C_3 = 10 \) | 10 |

---

📘 **Summary:**
- **Addition Principle:** Either–Or situations → add possibilities  
- **Multiplication Principle:** Sequential (And) situations → multiply possibilities  
- **Permutation:** Ordered arrangements  
- **Combination:** Unordered selections  

Together, they form the **foundation of probability and combinatorics**.


### 🎲 Applications of Permutations and Combinations in Probability

Permutations and Combinations are the foundation for calculating **probabilities**  
when events involve **arrangements** or **selections** of objects.

---

### 🧩 1. Relationship Between Counting and Probability

The **Probability of an Event (A)** is given by:

$$
P(A) = \frac{\text{Number of favourable outcomes}}{\text{Total number of possible outcomes}}
$$

To find these counts, we often use **permutation** or **combination** formulas.

---

## 🧮 2. Example 1 — Drawing Cards from a Deck

A standard deck has **52 cards**.

### 🎯 Example 1.1 — Probability of Getting an Ace

- Favourable outcomes = 4 Aces  
- Total outcomes = 52  

$$
P(\text{Ace}) = \frac{4}{52} = \frac{1}{13}
$$

✅ **Probability = 1/13**

---

### 🎯 Example 1.2 — Probability of Getting 2 Aces in 2 Draws (Without Replacement)

- Total ways to draw 2 cards:  
  \( {}^{52}C_2 = \frac{52!}{2!50!} = 1326 \)
- Favourable ways to get 2 Aces:  
  \( {}^4C_2 = 6 \)

$$
P(\text{2 Aces}) = \frac{{}^4C_2}{{}^{52}C_2} = \frac{6}{1326} = \frac{1}{221}
$$

✅ **Probability = 1/221**

---

## 🧮 3. Example 2 — Forming Words (Permutation in Probability)

> The letters of the word **APPLE** are arranged at random.  
> Find the probability that both P’s are together.

- Total letters = 5 (A, P, P, L, E)

### Step 1: Total arrangements
Since 2 P’s are identical:

$$
\text{Total arrangements} = \frac{5!}{2!} = 60
$$

### Step 2: Favourable arrangements (P’s together)
Treat “PP” as one letter → total letters = 4 (PP, A, L, E)

$$
\text{Favourable arrangements} = 4! = 24
$$

### Step 3: Probability

$$
P(\text{P’s together}) = \frac{24}{60} = \frac{2}{5}
$$

✅ **Probability = 2/5**

---

## 🧮 4. Example 3 — Selecting Students (Combination in Probability)

> A class has 5 boys and 3 girls.  
> If a committee of 3 is chosen at random,  
> find the probability that it includes **exactly 2 boys**.

- Total students = 8  
  ⇒ Total ways to choose 3 = \( {}^8C_3 = 56 \)
- Choose 2 boys out of 5 = \( {}^5C_2 = 10 \)
- Choose 1 girl out of 3 = \( {}^3C_1 = 3 \)
- Favourable = \( 10 \times 3 = 30 \)

$$
P(\text{2 boys, 1 girl}) = \frac{30}{56} = \frac{15}{28}
$$

✅ **Probability = 15/28**

---

## 🧮 5. Example 4 — Dice Problem (Counting Outcomes)

> Two dice are rolled. Find the probability of getting a sum of 7.

- Total outcomes = \( 6 \times 6 = 36 \)
- Favourable outcomes = (1,6), (2,5), (3,4), (4,3), (5,2), (6,1) = 6

$$
P(\text{Sum = 7}) = \frac{6}{36} = \frac{1}{6}
$$

✅ **Probability = 1/6**

---

## 🧮 6. Example 5 — Probability with Permutations (Arrangements)

> Find the probability that a 3-digit number formed from digits 1–5 (without repetition)  
> is **even**.

- Total 3-digit numbers = \( {}^5P_3 = 60 \)
- Even numbers → last digit must be even (2 or 4) → 2 choices  

For each:
- Fix last digit → remaining 2 digits = \( {}^4P_2 = 12 \)

Favourable = \( 2 \times 12 = 24 \)

$$
P(\text{Even number}) = \frac{24}{60} = \frac{2}{5}
$$

✅ **Probability = 2/5**

---

## 🧾 Summary Table

| Scenario | Concept Used | Formula | Result |
|-----------|---------------|----------|---------|
| Drawing an Ace | Combination | \( \frac{4}{52} \) | \( \frac{1}{13} \) |
| Drawing 2 Aces | Combination | \( \frac{{}^4C_2}{{}^{52}C_2} \) | \( \frac{1}{221} \) |
| P’s together (APPLE) | Permutation | \( \frac{4!}{5!/2!} \) | \( \frac{2}{5} \) |
| 2 Boys, 1 Girl | Combination | \( \frac{{}^5C_2 \times {}^3C_1}{{}^8C_3} \) | \( \frac{15}{28} \) |
| Sum = 7 on Dice | Multiplication Rule | \( \frac{6}{36} \) | \( \frac{1}{6} \) |
| 3-digit Even Number | Permutation | \( \frac{24}{60} \) | \( \frac{2}{5} \) |

---

### 💡 Key Takeaways

1. **Use Combinations** when **selection/order doesn’t matter**.  
2. **Use Permutations** when **order matters** (arrangements).  
3. Apply **Probability = Favourable / Total outcomes**.  
4. Many complex problems are built using these basic counting rules.  

---

| Question Type | Keyword | Concept |
|----------------|----------|----------|
| Choose / Select / Form | “OR” / “Of” | Combination |
| Arrange / Order / Seat | “AND” / “Sequence” | Permutation |

> **Addition + Multiplication + Permutation + Combination**


### 🎯 Conditional Probability and Bayes’ Theorem

---

### 🧩 1. Conditional Probability — Concept

Sometimes, we are interested in the probability of an event **given that** another event has already occurred.

This is known as **Conditional Probability**.

---

### 📘 Definition

The **conditional probability** of event \( A \) given that \( B \) has occurred is defined as:

$$
P(A|B) = \frac{P(A \cap B)}{P(B)}, \quad \text{where } P(B) > 0
$$

---

### 🧠 Interpretation

- \( P(A|B) \): Probability that **A occurs**, knowing that **B** has occurred.  
- The sample space is now **restricted to event B**.  
- Used when one event **depends** on another.

---

### 📊 Example 1 — Basic Conditional Probability

> A card is drawn from a deck.  
> Find the probability that it is a King, **given that** it is a face card.

- Total face cards = 12 (Jack, Queen, King in each suit)  
- Favourable outcomes (Kings) = 4  

$$
P(\text{King | Face}) = \frac{4}{12} = \frac{1}{3}
$$

✅ Probability = **1/3**

---

### 📊 Example 2 — Dependent Events (Without Replacement)

> A bag has 3 red and 2 blue balls.  
> Two balls are drawn one after another **without replacement**.  
> Find \( P(\text{Both Red}) \).

Step 1: \( P(\text{1st Red}) = \frac{3}{5} \)

Step 2: After drawing 1 red, remaining reds = 2, total = 4  
\( P(\text{2nd Red | 1st Red}) = \frac{2}{4} = \frac{1}{2} \)

Step 3: Multiply (Multiplication Rule for Dependent Events):

$$
P(\text{Both Red}) = P(\text{1st Red}) \times P(\text{2nd Red | 1st Red})
$$

$$
= \frac{3}{5} \times \frac{1}{2} = \frac{3}{10}
$$

✅ Probability = **3/10**

---

### 🧩 2. Multiplication Rule of Probability

Derived from the definition of conditional probability:

$$
P(A \cap B) = P(A|B) \times P(B)
$$

Or equivalently:

$$
P(A \cap B) = P(B|A) \times P(A)
$$

📘 Used to find the probability that **both A and B** occur.

---

### 🧩 3. Law of Total Probability

If events \( B_1, B_2, ..., B_n \) are **mutually exclusive and exhaustive**,  
and \( A \) is any event, then:

$$
P(A) = \sum_{i=1}^{n} P(B_i) \times P(A|B_i)
$$

📘 Used when event \( A \) can occur in **multiple disjoint ways**.

---

### 🧮 Example — Total Probability

> A student can choose one of three tutors (T1, T2, T3).  
> - P(T1) = 0.3, P(T2) = 0.4, P(T3) = 0.3  
> - Probability of passing given each tutor: P(Pass|T1)=0.9, P(Pass|T2)=0.8, P(Pass|T3)=0.7  
> Find P(Pass).

$$
P(Pass) = (0.3)(0.9) + (0.4)(0.8) + (0.3)(0.7)
$$

$$
P(Pass) = 0.27 + 0.32 + 0.21 = 0.80
$$

✅ Probability that the student passes = **0.8**

---

## 🧩 4. Bayes’ Theorem

### 📘 Statement

If \( B_1, B_2, ..., B_n \) form a **partition of the sample space**, and \( A \) is an event such that \( P(A) > 0 \),  
then the probability of \( B_j \) given \( A \) is:

$$
P(B_j|A) = \frac{P(B_j) \times P(A|B_j)}{\sum_{i=1}^{n} P(B_i) \times P(A|B_i)}
$$

---

### 🧠 Interpretation

Bayes’ theorem allows us to **reverse the condition** —  
it helps find \( P(\text{Cause | Effect}) \) using \( P(\text{Effect | Cause}) \).

Used widely in:
- Medical testing (disease vs test result)
- Machine learning (Naïve Bayes)
- Reliability and decision analysis

---

### 🧮 Example — Bayes’ Theorem

> A lab test for a disease is 98% accurate.  
> 1% of the population has the disease.  
> Find the probability that a person **actually has the disease** given that the test is positive.

Let:
- \( D \): Person has the disease  
- \( D' \): Person doesn’t have it  
- \( T \): Test is positive  

Given:
- \( P(D) = 0.01 \)
- \( P(T|D) = 0.98 \)
- \( P(T|D') = 0.02 \)

Find \( P(D|T) \).

By Bayes’ Theorem:

$$
P(D|T) = \frac{P(D)P(T|D)}{P(D)P(T|D) + P(D')P(T|D')}
$$

$$
= \frac{(0.01)(0.98)}{(0.01)(0.98) + (0.99)(0.02)} = \frac{0.0098}{0.0098 + 0.0198} = \frac{0.0098}{0.0296} \approx 0.331
$$

✅ **Probability = 0.33 (≈ 33%)**

📘 Meaning: Even if the test is positive, only 33% chance that the person actually has the disease — because the disease is rare.

---

## 🧾 Summary Table

| Concept | Formula | Used For |
|----------|----------|----------|
| Conditional Probability | \( P(A|B) = \frac{P(A \cap B)}{P(B)} \) | Probability of A given B |
| Multiplication Rule | \( P(A \cap B) = P(A|B)P(B) \) | Joint probability of A and B |
| Total Probability | \( P(A) = \sum P(B_i)P(A|B_i) \) | Event from multiple causes |
| Bayes’ Theorem | \( P(B_j|A) = \frac{P(B_j)P(A|B_j)}{\sum P(B_i)P(A|B_i)} \) | Reverse condition (effect → cause) |

---

### 💡 Quick Tips

1. **Conditional Probability** narrows the sample space.  
2. **Total Probability** breaks a problem into smaller disjoint parts.  
3. **Bayes’ Theorem** reverses conditional probability.  
4. **Independent Events:**  
   $$ P(A \cap B) = P(A)P(B) $$  
   and  
   $$ P(A|B) = P(A) $$

---

### 📘 Common Applications

| Scenario | Concept Used |
|-----------|---------------|
| Drawing cards/dice after one event | Conditional Probability |
| Machine learning classifiers | Bayes’ Theorem |
| Disease testing (true/false positive) | Bayes’ Theorem |
| Multi-step probability experiments | Multiplication Rule |
| Partitioned sample space | Total Probability |

---

✅ **In summary:**
Conditional and Bayes’ probabilities let us model **dependent events** —  
they are essential tools for **real-world uncertainty**, from diagnostics to data science.


### 🎯 Independent Events, Random Variables, Expected Value & Variance

---

## 🧩 1. Independent Events

### 📘 Definition
Two events **A** and **B** are said to be **independent** if the occurrence of one does **not affect** the occurrence of the other.

Mathematically:

$$
P(A \cap B) = P(A) \times P(B)
$$

or equivalently,

$$
P(A|B) = P(A) \quad \text{and} \quad P(B|A) = P(B)
$$

---

### 🧠 Example — Tossing Coins

> When two coins are tossed:
> - Let \( A \): getting Head on the first coin  
> - Let \( B \): getting Head on the second coin  

Both are independent since:

$$
P(A \cap B) = P(A)P(B) = \frac{1}{2} \times \frac{1}{2} = \frac{1}{4}
$$

✅ So, A and B are **independent events**.

---

### 💡 Note:
If events are **not independent**, they are said to be **dependent**, meaning:
$$
P(A|B) \neq P(A)
$$

---

## 🎲 2. Random Variables

### 📘 Definition
A **Random Variable (RV)** is a numerical variable that represents the outcome of a random experiment.

It can take **different values**, each with an associated probability.

---

### 🧩 Types of Random Variables

| Type | Description | Example |
|------|--------------|----------|
| **Discrete** | Takes countable values | # of heads in 3 coin tosses |
| **Continuous** | Takes infinite values in an interval | Height, time, temperature |

---

### 🧮 Example — Discrete Random Variable

> Toss a coin twice.  
> Let \( X \): number of heads obtained.

| Outcome | X | P(X) |
|----------|---|-------|
| TT | 0 | 1/4 |
| HT, TH | 1 | 2/4 |
| HH | 2 | 1/4 |

✅ \( X \) is a **discrete random variable**.

---

## 📊 3. Probability Distribution of a Random Variable

A **probability distribution** assigns probabilities to each possible value of the random variable \( X \).

It must satisfy:
$$
\sum P(X_i) = 1
$$

---

### Example — Distribution Table for Coin Toss

| X (Heads) | 0 | 1 | 2 |
|------------|---|---|---|
| P(X) | 1/4 | 1/2 | 1/4 |

✅ The sum of all probabilities = \( 1/4 + 1/2 + 1/4 = 1 \)

---

## 🧮 4. Expected Value (Mean) of a Random Variable

### 📘 Definition
The **Expected Value (E[X])** of a discrete random variable \( X \) is the **weighted average** of all possible values.

$$
E[X] = \sum X_i P(X_i)
$$

---

### 🧠 Example — Expected Number of Heads in 2 Tosses

| X | P(X) | X·P(X) |
|---|------|--------|
| 0 | 1/4 | 0 |
| 1 | 1/2 | 0.5 |
| 2 | 1/4 | 0.5 |

$$
E[X] = 0 + 0.5 + 0.5 = 1
$$

✅ Expected number of heads = **1**

---

## 📉 5. Variance and Standard Deviation of a Random Variable

### 📘 Definition

**Variance** measures how much the values of \( X \) deviate from the mean.

$$
Var(X) = E[(X - E[X])^2]
$$

Alternate (computational) formula:

$$
Var(X) = E[X^2] - [E[X]]^2
$$

**Standard Deviation** is the square root of variance:

$$
\sigma_X = \sqrt{Var(X)}
$$

---

### 🧮 Example — Variance of X (Coin Toss)

From before:

| X | P(X) | X² | X²·P(X) |
|---|------|----|----------|
| 0 | 1/4 | 0 | 0 |
| 1 | 1/2 | 1 | 0.5 |
| 2 | 1/4 | 4 | 1 |

$$
E[X^2] = 1.5, \quad E[X] = 1
$$

So:

$$
Var(X) = E[X^2] - [E[X]]^2 = 1.5 - 1^2 = 0.5
$$

and

$$
\sigma_X = \sqrt{0.5} = 0.707
$$

✅ **Variance = 0.5, Standard Deviation = 0.707**

---

## 🧾 6. Summary Table

| Concept | Formula | Meaning |
|----------|----------|----------|
| Independence | \( P(A \cap B) = P(A)P(B) \) | Events don’t affect each other |
| Expected Value | \( E[X] = \sum X_i P(X_i) \) | Mean of distribution |
| Variance | \( Var(X) = E[X^2] - [E[X]]^2 \) | Spread around mean |
| Std. Deviation | \( \sigma_X = \sqrt{Var(X)} \) | Measure of dispersion |

---

## 🧠 7. Important Properties

1. \( E[aX + b] = aE[X] + b \)
2. \( Var(aX + b) = a^2 Var(X) \)
3. If \( X, Y \) are independent:
   $$
   Var(X + Y) = Var(X) + Var(Y)
   $$

---

## 💡 Concept Summary

| Concept | Keyword | Example |
|----------|----------|----------|
| Independent Events | “No Influence” | Two coin tosses |
| Random Variable | “Numerical outcome” | Number of heads |
| Expected Value | “Average Outcome” | Mean number of successes |
| Variance | “Spread” | How far outcomes vary from mean |

---

✅ **In summary:**
- **Independent Events** → no dependence between occurrences  
- **Random Variables** → assign numerical outcomes  
- **Expected Value (E[X])** → average expected outcome  
- **Variance & σ** → measure of spread or uncertainty  

Together, these concepts are **core to Probability, Statistics, and Machine Learning foundations.**

# 🎯 DISCRETE PROBABILITY DISTRIBUTIONS

---

## 🧮 2. Binomial Distribution

### 📘 Definition
A **Binomial Distribution** gives the probability of getting exactly \( x \) successes  
in \( n \) independent trials, each with success probability \( p \).

### Formula

$$
P(X = x) = {n \choose x} p^x (1 - p)^{n - x}, \quad x = 0, 1, 2, ..., n
$$

where  
- \( n \): number of trials  
- \( p \): probability of success  
- \( q = 1 - p \): probability of failure

---

### 📊 Properties

| Property | Formula |
|-----------|----------|
| Mean | \( E[X] = np \) |
| Variance | \( Var(X) = np(1 - p) \) |
| Standard Deviation | \( \sigma = \sqrt{np(1 - p)} \) |

---

### 🧮 Example — Binomial Probability

> A coin is tossed 5 times. Find the probability of getting exactly 3 heads.

Here, \( n = 5, p = 0.5, q = 0.5 \)

$$
P(X = 3) = {5 \choose 3} (0.5)^3 (0.5)^2 = 10 \times 0.125 \times 0.25 = 0.3125
$$

✅ **P(X = 3) = 0.3125**

---

## 🧮 3. Poisson Distribution

### 📘 Definition
A **Poisson Distribution** gives the probability of a number of events  
occurring in a fixed interval (time, space, etc.) when the events occur independently and rarely.

It’s used as a **limiting case of the Binomial** when \( n \) is large and \( p \) is small.

### Formula

$$
P(X = x) = \frac{e^{-\lambda} \lambda^x}{x!}, \quad x = 0, 1, 2, ...
$$

where \( \lambda = np \) = mean number of occurrences.

---

### 📊 Properties

| Property | Formula |
|-----------|----------|
| Mean | \( E[X] = \lambda \) |
| Variance | \( Var(X) = \lambda \) |

---

### 🧮 Example — Poisson Probability

> The average number of cars passing through a toll gate per minute is 3.  
> Find the probability that exactly 4 cars pass in one minute.

Here, \( \lambda = 3, x = 4 \)

$$
P(X = 4) = \frac{e^{-3} 3^4}{4!} = \frac{e^{-3} \times 81}{24} \approx 0.168
$$

✅ **P(X = 4) ≈ 0.168**

# 🌈 CONTINUOUS PROBABILITY DISTRIBUTIONS

---

## 🧮 4. Normal Distribution

### 📘 Definition
A **Normal Distribution** is a continuous distribution that describes many natural phenomena  
(such as heights, weights, test scores, etc.).

It has the famous **bell-shaped curve**, symmetric about its mean \( \mu \).

---

### Formula

$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} \, e^{-\frac{(x - \mu)^2}{2\sigma^2}}, \quad -\infty < x < \infty
$$

where  
- \( \mu \): mean (center of the curve)  
- \( \sigma \): standard deviation (spread)

---

### 📊 Properties

| Property | Description |
|-----------|--------------|
| Shape | Symmetric bell curve |
| Mean = Median = Mode | \( \mu \) |
| Total Area under curve | 1 |
| Spread | Controlled by \( \sigma \) |

---

### 📘 Empirical Rule (68–95–99.7 Rule)

| Range | % of Data |
|--------|------------|
| \( \mu \pm 1\sigma \) | 68% |
| \( \mu \pm 2\sigma \) | 95% |
| \( \mu \pm 3\sigma \) | 99.7% |

✅ About 95% of data falls within 2 standard deviations of the mean.

---

## 🧮 5. Standard Normal Distribution

### 📘 Definition
A **Standard Normal Distribution** is a special case of the Normal distribution  
with mean \( \mu = 0 \) and standard deviation \( \sigma = 1 \).

The variable used is:

$$
Z = \frac{X - \mu}{\sigma}
$$

This converts any normal random variable \( X \) into a **Z-score**.

---

### 📊 Example — Finding a Z-Score

> A student scored 85 on a test where mean = 70 and SD = 10.  
> Find their Z-score.

$$
Z = \frac{X - \mu}{\sigma} = \frac{85 - 70}{10} = 1.5
$$

✅ The student scored **1.5 standard deviations above the mean**.

---

## 🧮 6. Relationship Between Distributions

| Distribution | Type | When Used | Parameters |
|---------------|------|------------|-------------|
| **Binomial** | Discrete | Fixed number of independent trials | \( n, p \) |
| **Poisson** | Discrete | Count of rare events per interval | \( \lambda \) |
| **Normal** | Continuous | Natural measurements and averages | \( \mu, \sigma \) |

---

## 📘 7. Connection Between Them

| Relationship | Meaning |
|---------------|----------|
| Poisson ≈ Binomial when \( n \) large, \( p \) small | \( \lambda = np \) |
| Normal ≈ Binomial when \( n \) large | Central Limit Theorem |

---

| Distribution | Formula | Mean | Variance | Type |
|---------------|----------|------|----------|------|
| Binomial | \( P(X=x) = {n \choose x} p^x (1-p)^{n-x} \) | \( np \) | \( np(1-p) \) | Discrete |
| Poisson | \( P(X=x) = \frac{e^{-\lambda} \lambda^x}{x!} \) | \( \lambda \) | \( \lambda \) | Discrete |
| Normal | \( f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} \) | \( \mu \) | \( \sigma^2 \) | Continuous |

### 💡 Key Takeaways

1. **Discrete** → countable outcomes (Binomial, Poisson)  
2. **Continuous** → continuous outcomes (Normal)  
3. **Mean (μ)** → central tendency  
4. **Variance (σ²)** → spread of distribution  
5. **Z-score** helps compare different normal distributions  
6. **Central Limit Theorem** → sums or averages of large samples ≈ Normal

---

✅ **In summary:**
- **Binomial** → Success/failure in fixed trials  
- **Poisson** → Rare events per interval  
- **Normal** → Continuous, natural data  
- **Standard Normal (Z)** → Normalized version for comparisons  

Together, these distributions form the **core framework of probability models in statistics and machine learning**.
