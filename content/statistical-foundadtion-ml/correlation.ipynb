#### Correlation

In Statistics, Correlation measures the strength and direction of the relationship between two numerical variables.

#### Types of Correlation

**Positive Correlation**  ->	Both variables increase together	Height & Weight

**Negative Correlation** ->	One increases, the other decreases	Speed & Travel Time

**Zero Correlation** ->	No relationship	Shoe size & Intelligence



#### Mathematical Measure: Pearson’s Correlation Coefficient (r)

In terms of covariance and standard deviations:

$$
r = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y}
$$

Where:

Cov(X, Y) = covariance between X and Y

σX, σY = standard deviations of X and Y

The value of r always lies between -1 and +1:


#### The formula for Pearson’s correlation coefficient is:

$$
r = \frac{n(\sum XY) - (\sum X)(\sum Y)}{\sqrt{[n\sum X^2 - (\sum X)^2][n\sum Y^2 - (\sum Y)^2]}}
$$

#### Covariance Formula

$$
\text{Cov}(X, Y) = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{n - 1}
$$

#### Population Covariance
If we dealing with the entire population (not just a sample), use n instead of 𝑛−1


$$
\text{Cov}(X, Y) = \frac{\sum_{i=1}^{n} (X_i - \mu_X)(Y_i - \mu_Y)}{n}
$$
 


```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------
# Data: Hours Studied vs Marks Scored
# ---------------------------------
X = np.array([2, 4, 6, 8, 10])
Y = np.array([40, 50, 60, 70, 80])

# ---------------------------------
# Calculate correlation coefficient
# ---------------------------------
r = np.corrcoef(X, Y)[0, 1]

# ---------------------------------
# Plot scatter plot
# ---------------------------------
plt.scatter(X, Y, s=80)
plt.plot(X, Y, linestyle='--', label=f'Correlation (r) = {r:.2f}')
plt.xlabel('Hours Studied (X)')
plt.ylabel('Marks Scored (Y)')
plt.title('Correlation between Study Hours and Marks')
plt.legend()
plt.grid(True)
plt.show()
```
