#### The Empirical Rule

The Empirical Rule (also known as the 68–95–99.7 Rule) is a fundamental concept in statistics that describes how data is distributed in a normal distribution (bell-shaped curve).

The Empirical Rule states that for a normal distribution:

**μ ± 1σ	Within 1 standard deviation of the mean	   68%**

**μ ± 2σ	Within 2 standard deviations of the mean   95%**

**μ ± 3σ	Within 3 standard deviations of the mean   99.7%**

Where:

μ (mu) = mean of the data

σ (sigma) = standard deviation



#### Use Cases:

Quickly estimate probabilities for normally distributed data

Detect outliers (values beyond 3σ are rare)

Useful for quality control, finance, psychometrics, and machine learning feature analysis



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# -------------------------------
# Dataset and basic statistics
# -------------------------------
scores = np.array([55, 60, 62, 63, 64, 65, 66, 67, 68, 69, 
                   70, 71, 72, 73, 74, 75, 76, 77, 78, 80, 
                   82, 84, 86, 88, 90])

mean = np.mean(scores)
std = np.std(scores, ddof=1)

# -------------------------------
# Plot normal distribution
# -------------------------------
x = np.linspace(40, 100, 500)
y = norm.pdf(x, mean, std)

plt.figure(figsize=(10,6))
plt.plot(x, y, label='Normal Distribution', linewidth=2)

# Shade 1σ, 2σ, and 3σ regions
for i, color in zip(range(1, 4), ['#a1d99b', '#74c476', '#31a354']):
    plt.fill_between(x, y, 0, where=(x > mean - i*std) & (x < mean + i*std), color=color, alpha=0.4,
                     label=f'μ ± {i}σ')

# Mean line
plt.axvline(mean, color='black', linestyle='--', label='Mean (μ)')

plt.title("Empirical Rule Visualization (68–95–99.7%)")
plt.xlabel("Scores")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True)
plt.show()

mean, std
```
