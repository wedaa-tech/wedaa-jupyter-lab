
#### Variance and standard deviation:

Variance and standard deviation both measure the spread of data around the mean, but standard deviation is the square root of the variance. This means variance is in squared units, while standard deviation is in the same units as the original data, making it easier to interpret. Variance is the average of the squared differences from the mean, and standard deviation is the square root of this value. 

#### Standard Deviation
Standard Deviation (σ or s) measures how spread out the data values are from the mean (average).
It tells you how much the data varies.

For sample data:

$$
s = \sqrt{\frac{\sum_{i=1}^{n} (X_i - \bar{X})^2}{n - 1}}
$$


For a Population

$$
\sigma = \sqrt{\frac{\sum_{i=1}^{n} (X_i - \mu)^2}{n}}
$$


Interpretation:

A smaller SD → data points are close to the mean.

A larger SD → data points are more spread out.


```python
import numpy as np

# Example data: Hours studied by 4 students
X = np.array([1, 2, 3, 4])

# Calculate Mean
mean_X = np.mean(X)

# Calculate Standard Deviation (Population)
std_population = np.std(X)

# Calculate Standard Deviation (Sample)
std_sample = np.std(X, ddof=1)  # ddof=1 → n-1 in denominator

print("Data:", X)
print("Mean:", mean_X)
print("Population Std Dev (σ):", std_population)
print("Sample Std Dev (s):", std_sample)
```
